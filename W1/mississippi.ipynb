## Mississippi, in Block Letters

Another early exercise from COMP 170: write a program that prints the name of the longest river in the US, in block letters. The version you likely wrote back then prints it *vertically*, one letter at a time, top to bottom.

```python
def m():
    print(" M       M")
    print(" M M   M M")
    print(" M  M M  M")
    print(" M   M   M")
    print(" M       M")
    print(" M       M")
    print()


def i():
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print()


def s():
    print("   SSSSSSS")
    print(" SS       ")
    print(" SS       ")
    print("  SSSSSS  ")
    print("        SS")
    print("        SS")
    print(" SSSSSSS  ")
    print()


def p():
    print(" PPPPPPPP ")
    print(" P      PP")
    print(" P     PP ")
    print(" PPPPPP   ")
    print(" P        ")
    print(" P        ")
    print(" P        ")
    print()


def ssi():
    s()
    s()
    i()


def ssissi():
    ssi()
    ssi()


def pp():
    p()
    p()


def mississippi():
    m()
    i()
    ssissi()
    pp()
    i()


def main():
    mississippi()


main()
```

A nice variation on this problem is to print the name horizontally instead. To do that, we first need to stop thinking of each letter as *something we print* and start thinking of it as *something we have* — data we can look at, pass around, and combine, rather than a sequence of `print` calls locked inside a function.

Here are the same four letters, rewritten as lists of strings. Each list holds the letter's rows, top to bottom.

```python
M = [
    " M       M",
    " M M   M M",
    " M  M M  M",
    " M   M   M",
    " M       M",
    " M       M",
    " M       M",
]

I = [
    "     I    ",
    "     I    ",
    "     I    ",
    "     I    ",
    "     I    ",
    "     I    ",
    "     I    ",
]

S = [
    "   SSSSSSS",
    " SS       ",
    " SS       ",
    "  SSSSSS  ",
    "        SS",
    "        SS",
    " SSSSSSS  ",
]

P = [
    " PPPPPPPP ",
    " P      PP",
    " P     PP ",
    " PPPPPP   ",
    " P        ",
    " P        ",
    " P        "
]
```

Using this abstraction, the vertical printing can be accomplished as follows. Notice that neither function below knows anything about M, I, S, or P specifically — they just work through whatever list of letters they're handed.

```python
def print_vertical(block_letter):
    for line in block_letter:
        print(line)
    print()


def compose_vertical(letters):
    for letter in letters:
        print_vertical(letter)


big_river = [M, I, S, S, I, S, S, I, P, I]
compose_vertical(big_river)
```

But we're not interested in vertical printing anymore. We want to display the name horizontally — which means the loop has to change shape. Instead of finishing one whole letter before starting the next, we now need to print *line 1 of every letter*, then *line 2 of every letter*, and so on — the outer and inner loop swap roles.

```python
def compose_horizontal(letters):
    # Assume that each letter has the same height
    height = len(letters[0])
    # For each line from top to bottom of a block letter
    for line in range(height):
        # print that line for each letter
        for letter in letters:
            # Do not break line for each letter (end="")
            print(letter[line], end="")
        # Break a line after each letter has been processed at this line level
        print()


big_river = [M, I, S, S, I, S, S, I, P, I]
compose_horizontal(big_river)
```

The output isn't very pleasant — partly because our block letters aren't neatly designed, and partly because we're printing monospaced letters and it shows. Both are opportunities to improve the code a bit. All it takes, really, is less empty space around the letter `I`, and because `I` is *data* now, fixing it means editing a list — not rewriting a function.

```python
I = [
    "  I ",
    "  I ",
    "  I ",
    "  I ",
    "  I ",
    "  I ",
    "  I ",
]

big_river = [M, I, S, S, I, S, S, I, P, I]
compose_horizontal(big_river)
```

---

### Composing vs. coding the composition

Look back at `mississippi()` in the very first cell: `m(); i(); ssissi(); pp(); i()` — a sequence hand-written for exactly the word "Mississippi." Now compare it to `big_river = [M, I, S, S, I, S, S, I, P, I]` plus `compose_horizontal(big_river)`.

Suppose we wanted to print "Ohio" instead (imagine `O` and `H` shapes already existed).

| Approach | What changes to print "Ohio" instead |
|---|---|
| `mississippi()`-style, one function per word | A brand new function, hand-sequencing `o()`, `h()`, `i()`, `o()` |
| List + `compose_horizontal` | One new list: `["O", "H", "I", "O"]` looked up against existing letter shapes — `compose_horizontal` itself doesn't change |

Once a letter is data instead of a function call, *printing a different word* stops being a coding problem and becomes a data problem. That's the same shift we saw with the nursery rhyme last class, showing up again in a completely different problem.

### One thing we never actually coded

We only ever defined shapes for four letters: `M`, `I`, `S`, `P` — exactly the ones "Mississippi" happens to need. What would it take to print *any* river name this way — "Ohio," say, or "Colorado"?

You'd need a shape for every letter that could show up, not just the four we happened to need today. And once you're storing 26 letters instead of 4, indexing them by *name* (`M`, `I`, `S`, ...) rather than by position starts to matter a lot. That's not a coincidence — it's close to the exact problem Week 4 picks up.
