**Math and notation, from COMP 163**

Data structures and algorithm analysis lean on notation from COMP 163: number sets, quantifiers, set operations, and asymptotic notation. Before Week 01, review:

- **Number sets:** $\mathbb N$, $\mathbb R$, and subset notation, for example $\mathbb R_{<0}$
- **Function declarations:** $f:\mathbb N\rightarrow \mathbb R$
- **Quantifiers:** $\exists$, $\forall$
- **Set membership:** $c\in\mathbb R_{>0}$
- **Set operations and relations:** $\cup$, $\cap$, $\times$, $\subset$, $\supset$
- **Important functions:** $\log_2n$, $e^n$, $n!$, $\binom{n}{k}$

With a bit of review, you should be able to read, decode, and understand expressions such as the formal definition of Big-O:

$$
\begin{aligned}
&\text{Let } f,g : \mathbb{N} \to \mathbb{R}. \\
&g(n) \in \mathcal{O}(f(n))
\;\Longleftrightarrow\;
\exists\, c \in \mathbb{R}_{>0},\ \exists\, n_0 \in \mathbb{N}
\text{ such that} \\
&\forall\, n \in \mathbb{N},\ n \ge n_0,\ g(n) \le cf(n).
\end{aligned}
$$

#### Step by step: getting started with git from the terminal

This is the workflow I'd like everyone comfortable with by the end of the semester, and there's no reason not to start now:

1. **Check git is installed:** run `git --version` in your terminal. It's preinstalled on most Mac/Linux systems; on Windows, install [Git for Windows](https://git-scm.com/download/win), which also gives you Git Bash.
2. **Tell git who you are** (one-time setup):
   ```
   git config --global user.name "Your Name"
   git config --global user.email "you@example.com"
   ```
3. **Clone the course repo:**
   ```
   git clone <course-repo-url>
   ```
   This downloads the whole repo to your machine.
4. **Move into a week's folder:**
   ```
   cd course-repo-name/Week01
   ```
5. **Pull new material as it's added** throughout the semester:
   ```
   git pull
   ```
6. **For your own coursework**, create your own repository on github.com ("New repository"), then connect a local folder to it:
   ```
   git init
   git remote add origin <your-repo-url>
   git add .
   git commit -m "Week 1 work"
   git push -u origin main
   ```

> **Why keep your own work on GitHub, not just on your machine?** Because your code lives on GitHub, you can open a Codespace against it from *any* machine you log into: your laptop dies, you're on a lab computer, whatever, and pick up exactly where you left off, no setup required. Think of Codespaces as that hassle-free, portable safety net, not your day-to-day development environment. The real goal this semester is to be comfortable in an actual local IDE with the terminal; Codespaces is there for when you need it, not a substitute for learning that.
