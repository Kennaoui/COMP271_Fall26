# Type Annotations

## COMP 271 — Data Structures I

Type annotations make the expected inputs, outputs, and collection contents visible before we read a function body.

## Learning goals

By the end of this lesson, you should be able to:

- explain why Python is dynamically yet strongly typed;
- annotate variables, parameters, and return values;
- describe collection contents with annotations such as `list[str]` and `list[list[int]]`;
- use `type | None` for a result that may be absent; and
- explain the relationship between annotations, Python, and `mypy`.

## Dynamically typed and strongly typed

Python is **dynamically typed**: types are checked while the program runs, and a name may later refer to a value of another type.

```python
x = 42
x = "now a string"       # legal in Python
```

Python is also **strongly typed**: it usually refuses to silently combine unrelated types.

```python
3 + "4"                  # TypeError
```

Dynamic typing does not mean that values have no types. It means that variable declarations do not permanently restrict names to one type.

## What a type annotation looks like

An annotation records the type we expect.

```python
count: int = 0
names: list[str] = ["Ana", "Raj"]

def greet(name: str, times: int = 1) -> str:
    return ("Hello " + name + "! ") * times
```

In a function signature:

- annotations after parameter names describe expected arguments;
- the annotation after `->` describes the return value;
- `-> None` means the function does not return a useful value.

````python
def greet(name: str, times: int = 1) -> str:
    """Build and return a repeated greeting."""
    greeting = ("Hello " + name + "! ") * times
    return greeting


print(greet("Ana", 2))
print(greet.__annotations__)
````

## Why annotate code?

- **Readability:** the function contract is visible without reading its body.
- **Tool support:** editors can offer better completion and warnings.
- **Earlier feedback:** a static checker can find inconsistent uses before that path executes.
- **Teamwork:** callers can understand a function written by someone else.

Annotations are especially valuable at function boundaries. Local variables often remain clear from context.

## Collections describe two levels of structure

Compare:

```python
stations: list[str]
grid: list[list[int]]
scores: dict[str, float]
```

- `list[str]` means a list whose elements are strings.
- `list[list[int]]` means a list whose elements are lists of integers.
- `dict[str, float]` means string keys mapped to floating-point values.

For example, `list[list[int]]` can represent a grid: the outer list contains rows, and each row is a list of integers.

## When a result may be missing

Use `type | None` when a function may return either a value or no value.

```python
def find_student(student_id: int) -> str | None:
    return roster.get(student_id)
```

The annotation tells the caller to handle both possibilities:

```python
name = find_student(42)
if name is None:
    print("No such student")
else:
    print(name)
```

`None` is a real value representing the absence of a result; it is not the same as `""`, `0`, or `False`.

## Hints, not runtime guarantees

Python normally stores annotations but does not enforce them while the program runs.

```python
def double(x: int) -> int:
    return x * 2

double("ab")             # produces "abab"
```

A separate static checker, such as `mypy`, analyzes the annotations:

```bash
mypy program.py
```

A correct annotation documents a promise. An incorrect annotation is worse than no annotation because it misleads both people and tools.

## Practical annotation guidelines

- Start with function parameters and return types.
- Describe container contents: prefer `list[str]` to plain `list`.
- Prefer a specific type to `Any`.
- Use built-in generic forms such as `list[str]`, not the older `List[str]` spelling.
- Keep annotations readable; an extremely complicated type may reveal an overly complicated design.
- Update annotations whenever behavior changes.

## Small practice

Add accurate annotations to this function:

```python
def average(scores):
    total = sum(scores)
    return total / len(scores)
```

Then consider what should happen if `scores` is empty. Type annotations describe expected types, but they do not solve that behavioral question for us.
