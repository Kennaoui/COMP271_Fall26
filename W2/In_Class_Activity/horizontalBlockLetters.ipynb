## Mississippi, in Block Letters

Another early exercise from COMP 170: write a program that prints the name of the longest river in the US, in block letters. The version you likely wrote back then prints it *vertically*, one letter at a time, top to bottom.

```python
def m():
    print(" M       M")
    print(" M M   M M")
    print(" M  M M  M")
    print(" M   M   M")
    print(" M       M")
    print(" M       M")
    print()

def i():
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print("     I    ")
    print()

def s():
    print("   SSSSSSS")
    print(" SS       ")
    print(" SS       ")
    print("  SSSSSS  ")
    print("        SS")
    print("        SS")
    print(" SSSSSSS  ")
    print()

def p():
    print(" PPPPPPPP ")
    print(" P      PP")
    print(" P     PP ")
    print(" PPPPPP   ")
    print(" P        ")
    print(" P        ")
    print(" P        ")
    print()

def ssi():
    s()
    s()
    i()

def ssissi():
    ssi()
    ssi()

def pp():
    p()
    p()

def mississippi():
    m()
    i()
    ssissi()
    pp()
    i()

def main():
    mississippi()

main()
```

A nice variation on this problem is to print the name horizontally instead. To do that, we first need to stop thinking of each letter as *something we print* and start thinking of it as *something we have* — data we can look at, pass around, and combine, rather than a sequence of `print` calls locked inside a function.

The starter file `blockLetters.py` contains the four letters `M`, `I`, `S`, and `P`, rewritten as lists of strings. Each list holds the letter's rows, top to bottom. It also contains a `sanity_check` function that checks that the letter data is well formed, including that every letter has the same height.

Using this abstraction, the vertical printing can be accomplished as follows. Notice that neither function below knows anything about M, I, S, or P specifically — they just work through whatever list of letters they're handed.

```python
def print_vertical(block_letter):
    for line in block_letter:
        print(line)
    print()

def compose_vertical(letters):
    for letter in letters:
        print_vertical(letter)

big_river = [M, I, S, S, I, S, S, I, P, P, I]
compose_vertical(big_river)
```
